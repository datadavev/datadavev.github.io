Semantic similarity of words.

Can we arrange the controlled vocabs a bit using a bit of assistance form wordnet?

```python
source_term_text = """amber
anthracite
antler
bars (building materials)
basalt
bone
bone (material)
brass (alloy)
brick
brick (clay product)
bronze
Bucchero
cast iron
ceramic
ceramic (material)
charcoal (material)
chert
cinder
clay
coal
Cobalt glass
copper alloy
dolomite
faience (composite material)
flint (rock)
gabbro
glass
glass (material)
gold
greywacke
hair
hematite
horn
iron
iron (metal)
ivory
kaolin"""
term_rows = [v.strip() for v in source_term_text.split("\n")]
term_words = [w[0].lower() for w in [row.split(" ") for row in term_rows]]
print(term_words)
```

```python
from nltk.corpus import wordnet as wn
from nltk.corpus import wordnet_ic
brown_ic = wordnet_ic.ic('ic-brown.dat')
#print(brown_ic)
algorithms = [
    ["path_similarity",], 
    ["lch_similarity",], 
    ["wup_similarity",],
    ["res_similarity", brown_ic],
    ["jcn_similarity", brown_ic],
]

a = wn.synset("dog.n.01")
print(a.__class__)
b = wn.synset("cat.n.01")
#for algo in algorithms:
#    params = [b, ]
#    if len(algo) > 1:
#        params = params + algo[1:]
#    print(f"{algo[0]}: {getattr(a, algo[0])(*params)}")

print(b.lin_similarity(a, brown_ic))
print(a)
print(b)
```

```python
term_words = ["ice", "water", ]
for word in term_words:
    ss = wn.synsets(word, pos=wn.NOUN)
    print(word)
    for s in ss:
        print(f"  {s.name()}")
        print(f"    {', '.join(s.lemma_names())}")
        print(f"    {s.definition()}")
```
