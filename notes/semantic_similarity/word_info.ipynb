```python
from nltk.corpus import wordnet as wn
import nltk.corpus.reader.wordnet
from nltk.corpus import wordnet_ic
brown_ic = wordnet_ic.ic('ic-brown.dat')


def tsyn(s):
    res = [
        s.name(),
        f"  {', '.join(s.lemma_names())}",
        f"  {s.definition()}",
    ]
    return "\n".join(res)

def syn_broader(s, depth=1):
    hyper = lambda s: s.hypernyms()
    for t in s.closure(hyper, depth=depth):
        yield t

def syn_narrower(s, depth=1):
    hyper = lambda s: s.hyponyms()
    for t in s.closure(hyper, depth=depth):
        yield t  
        
def syn_d(a, b):
    return a.path_similarity(b)
```

```python
term = wn.synset("water.n.01")
print(tsyn(term))
for t in syn_broader(term):
    print(f"{t.name():25} {syn_d(term, t):.3f}")
```

```python
for t in syn_narrower(term, depth=1):
    print(f"{syn_d(term, t):6.3f} {tsyn(t)}\n---")
```
